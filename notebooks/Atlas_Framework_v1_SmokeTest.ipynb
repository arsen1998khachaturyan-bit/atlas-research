# Atlas Research Framework v1.0 — Smoke Test

Загрузите `atlas-research-framework-v1.zip`, затем выполните все ячейки.


```python

import sys, subprocess, zipfile, shutil, os
from pathlib import Path
from google.colab import files

uploaded = files.upload()
zip_name = next(iter(uploaded))

extract_root = Path("/content/atlas_framework")
shutil.rmtree(extract_root, ignore_errors=True)
extract_root.mkdir(parents=True)

with zipfile.ZipFile(zip_name, "r") as zf:
    zf.extractall(extract_root)

project_roots = list(extract_root.rglob("pyproject.toml"))
if not project_roots:
    raise FileNotFoundError("pyproject.toml not found")

project_root = project_roots[0].parent
os.chdir(project_root)

subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", ".[dev]"], check=True)
subprocess.run([sys.executable, "-m", "pytest", "-q"], check=True)
subprocess.run([sys.executable, "-m", "experiments.run_smoke"], check=True)

archive = shutil.make_archive(
    "/content/atlas_framework_smoke_results",
    "zip",
    project_root / "results",
)
files.download(archive)
```
